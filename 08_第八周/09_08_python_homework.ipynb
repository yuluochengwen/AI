#### PAT-1106  2019数列
1. 创建**初始数列**，存储2019四个元素。
1. 编写一个函数```new_num```，计算数列**指定索引位置之前**的**四个元素的和**，用```%```运算来取和的个位数，将这个数添加到初始数列后面。
3. 编写循环，**从索引4开始**，循环调用函数。
4. 输出时要注意，N的范围是[0,1000]，所以**当N小于初始列表的长度时**，我们要进行列表切片。

````python
N = int(input())
result = [2, 0, 1, 9]
def new_num(x):
    sum_top4 = sum(result[x-4:x])
    num = sum_top4 % 10
    result.append(num)

for i in range(4, N):
    new_num(i)

print(''.join(map(str, result[0:N])))
````

#### PAT-1108 String复读机

````python
input_str = input()
cha_num = [0]*6
for ch in input_str:
    if ch == 'S':
        cha_num[0] += 1
    elif ch == 't':
        cha_num[1] += 1
    elif ch == 'r':
        cha_num[2] += 1
    elif ch == 'i':
        cha_num[3] += 1
    elif ch == 'n':
        cha_num[4] += 1
    elif ch == 'g':
        cha_num[5] += 1

while sum(cha_num) != 0:
    for i in range(6):
        if cha_num[i] != 0:
            if i == 0:
                print('S', end='')
            elif i == 1:
                print('t', end='')
            elif i == 2:
                print('r', end='')
            elif i == 3:
                print('i', end='')
            elif i == 4:
                print('n', end='')
            elif i == 5:
                print('g', end='')
            cha_num[i] -= 1
````

````python
# PTA-1108
input_str = input()

# 统计字符频次
counts = {ch: input_str.count(ch) for ch in "String"}

# 循环输出
while any(counts[ch] > 0 for ch in "String"):
    for ch in "String":
        if counts[ch] > 0:
            print(ch, end="")
            counts[ch] -= 1
````
