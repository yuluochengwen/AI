### torch.nn

````python
import torch
import torch.nn as nn

# 定义一个简单的神经网络模型
class SimpleNN(nn.Module):
    def __init__(self):
        super(SimpleNN, self).__init__()
        # 定义一个输入层到隐藏层的全连接层
        self.fc1 = nn.Linear(2, 2)  # 输入 2 个特征，输出 2 个特征
        # 定义一个隐藏层到输出层的全连接层
        self.fc2 = nn.Linear(2, 1)  # 输入 2 个特征，输出 1 个预测值
   
    def forward(self, x):
        # 前向传播过程
        x = torch.relu(self.fc1(x))  # 使用 ReLU 激活函数
        x = self.fc2(x)  # 输出层
        return x

# 创建模型实例
model = SimpleNN()

# 打印模型
print(model)
````

## 回归

````python
import numpy as np
import torch.nn as nn
import random
import matplotlib.pyplot as plt

x = np.arange(20)
print(x)
y = np.array([5*x[i] + random.randint(1, 20) for i in range(len(x))])

plt.scatter(x, y)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Scatter plot of y vs x')
plt.show()
````

````python
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams["axes.unicode_minus"] = False  # 正确显示负号

x_train = torch.from_numpy(x).float()
y_train = torch.from_numpy(y).float()
````

````python

class LinearRegressionModel(nn.Module):
    def __init__(self):
        super(LinearRegressionModel, self).__init__()
        self.fc1 = nn.Linear(1, 16)    # 输入层 -> 隐藏层1
        self.fc2 = nn.Linear(16, 8)    # 隐藏层1 -> 隐藏层2  
        self.fc3 = nn.Linear(8, 4)     # 隐藏层2 -> 隐藏层3
        self.fc4 = nn.Linear(4, 1)     # 隐藏层3 -> 输出层
        
    def forward(self, x):
        x = torch.relu(self.fc1(x))
        x = torch.relu(self.fc2(x))
        x = torch.relu(self.fc3(x))
        x = self.fc4(x)  # 输出层通常不加激活函数
        return x

model = LinearRegressionModel()

criterion = torch.nn.MSELoss()  # 均方误差损失函数
optimizer = torch.optim.SGD(model.parameters(), lr=0.0001)  # 随机梯度下降优化器


num_epochs = 100000
plot_epochs = []  # 存储要绘制的轮次
predictions = []  # 存储对应的预测结果
losses_for_plot = []  # 存储对应的损失值

for i in range(num_epochs):
    input_data = x_train.unsqueeze(1) # 调整输入形状为 (batch_size, 1)
    target = y_train.unsqueeze(1) # 调整目标形状为 (batch_size, 1)

    out = model(input_data) # 前向传播
    loss = criterion(out, target)  # 计算损失
    optimizer.zero_grad()  # 清零梯度
    loss.backward()  # 反向传播
    optimizer.step()  # 更新参数

    if i % 10000 == 0:  # 每10000轮打印一次
        print(f'Epoch [{i+1:6d}/{num_epochs}], Loss: {loss.item():.6f}')

    # 每10000轮记录一次绘图数据（减少子图数量）
    if i % 10000 == 0:
        with torch.no_grad():
            predict = model(input_data)
            plot_epochs.append(i+1)
            predictions.append(predict.squeeze().data.numpy().copy())
            losses_for_plot.append(loss.item())

# 绘制所有训练过程在一个画布中
n_plots = len(plot_epochs)
cols = 5  # 每行5个子图
rows = (n_plots + cols - 1) // cols  # 计算需要的行数

plt.figure(figsize=(15, 3*rows))  # 固定每个子图的尺寸比例

for idx, (epoch, pred, loss_val) in enumerate(zip(plot_epochs, predictions, losses_for_plot)):
    plt.subplot(rows, cols, idx + 1)
    
    # 绘制真实数据点
    plt.scatter(x_train.data.numpy(), y_train.data.numpy(), alpha=0.6, s=20, c='blue', label='真实数据')
    
    # 绘制预测线
    plt.plot(x_train.data.numpy(), pred, 'r-', linewidth=2, label='预测线')
    
    plt.xlabel('x', fontsize=10)
    plt.ylabel('y', fontsize=10)
    plt.title(f'Epoch {epoch}\nLoss: {loss_val:.4f}', fontsize=10)
    plt.grid(True, alpha=0.3)
    
    # 设置相同的坐标轴范围，确保单位长度相同
    plt.xlim(x_train.min().item() - 1, x_train.max().item() + 1)
    plt.ylim(y_train.min().item() - 5, y_train.max().item() + 5)
    
    # 只在第一个子图显示图例
    if idx == 0:
        plt.legend(fontsize=8)

plt.tight_layout()
plt.show()
````

### 生成数据

````python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from torch import optim

# 生成二分类数据
torch.manual_seed(1)
x1 = torch.rand(100)*2
y1 = torch.rand(100)*2
data = list(zip(x1, y1))


pos = []
neg = []

def classification(data):
    """
    分类函数
    """
    for i in data:
        if (i[0] - i[1]) < 0:
            pos.append(i)
        else:
            neg.append(i)

classification(data)

pos_x = [i[0] for i in pos]
pos_y = [i[1] for i in pos]
neg_x = [i[0] for i in neg] 
neg_y = [i[1] for i in neg]

plt.scatter(pos_x, pos_y, alpha=0.6, s=15, c='b', label='正样本')
plt.scatter(neg_x, neg_y, alpha=0.6, s=15, c='r', label='负样本')
plt.xlabel('x')
plt.ylabel('y')
plt.title('二分类数据可视化')
plt.legend()
plt.show()
````

### 构造特征和目标

````python

x_data = [[i[0], i[1]] for i in pos]  # 特征
x_data.extend([[i[0], i[1]] for i in neg]) # 合并,扩展
x_data = torch.Tensor(x_data) # 转为浮点型张量
y_data = [1 for _ in range(len(pos))]  # 标签
y_data.extend([0 for _ in range(len(neg))]) # 合并,扩展
y_data = torch.Tensor(y_data).view(-1, 1) # 调整形状为 (batch_size, 1),view()视图
````

### 搭建模型并训练

````python

# 搭建二分类模型
class LogisticRegressionModel(nn.Module):
    def __init__(self):
        super(LogisticRegressionModel, self).__init__()
        self.linear = nn.Linear(2, 1) # 输入2个特征，输出1个预测值
        self.sigmoid = nn.Sigmoid()  # Sigmoid 激活函数
        
    def forward(self, x):
        x = self.linear(x)
        x = self.sigmoid(x)  # 使用Sigmoid激活函数
        return x
    
model = LogisticRegressionModel()
criterion = nn.BCELoss()  # 二分类交叉熵损失函数
optimizer = optim.SGD(model.parameters(), lr=0.1)  # 随机梯度下降优化器

for epoch in range(10000):
    y_pred = model(x_data)  # 前向传播
    loss = criterion(y_pred, y_data)  # 计算损失
    if epoch % 1000 == 0:
        print(f'Epoch [{epoch}/10000], Loss: {loss.item():.4f}')
    optimizer.zero_grad()  # 清零梯度
    loss.backward()  # 反向传播
    optimizer.step()  # 更新参数
````

### 画决策边界

````python
w = model.linear.weight.data[0]
w0 = w[0]
w1 = w[1]
b = model.linear.bias.item()
print(w)
print(b)

with torch.no_grad():
    x = torch.arange(0,3).view(-1,1)
    y = (-w0*x - b)/w1  # 计算决策边界的y值
    plt.plot(x.numpy(),y.numpy())

plt.scatter(pos_x,pos_y,c='b',label='正样本')
plt.scatter(neg_x,neg_y,c='r',label='负样本')
plt.xlim(0,2)
plt.ylim(0,2)
plt.show()
````
