```python
import pandas as pd
from collections import Counter
from sklearn.datasets import make_classification
from imblearn.over_sampling import SMOTENC

# 1. 构造一个既有数值型又有类别型的样本数据
X, y = make_classification(n_samples=200, n_features=5, 
                           n_informative=3, n_redundant=0,
                           weights=[0.8, 0.2], random_state=42)

# 假设第2列和第4列是类别型变量（用0/1表示）
X = pd.DataFrame(X, columns=[f"f{i}" for i in range(5)])
X['f1'] = (X['f1'] > 0).astype(int)   # 类别型
X['f3'] = (X['f3'] > 0).astype(int)   # 类别型

print("采样前类别分布:", Counter(y))

# 2. 定义类别型特征的索引
categorical_features = [1, 3]

# 3. 使用SMOTENC过采样
smote_nc = SMOTENC(categorical_features=categorical_features, random_state=42)
X_res, y_res = smote_nc.fit_resample(X, y)

print("采样后类别分布:", Counter(y_res))
```
