```python
input = make_input(
"""
12
23 16 87 233 87 16 87 233 23 87 233 16
""")

N = int(input())
input_array = list(map(int, input().split()))
nums_count = {}
for i in input_array:
    if i in nums_count:
        nums_count[i] += 1
    else:
        nums_count[i] = 1

print(nums_count)
ans = 0
for key, value in nums_count.items():
    if value % 2 == 1 and key % 2 != 0:
        ans = key
        break
print(ans)
```

#### PAT-1123 舍入

```python
from decimal import *
 
 
# 获取输入数据
n, d = map(int, input().split())  # 获取首行输入
 
# 处理结果并输出
context = getcontext()  # 获取 decimal 模块的设置
context.prec = 300  # 设置 decimal 精度
for _ in range(n):  # 遍历行
    c, num = input().split()  # 获取该行的输入数据
    context.rounding = [ROUND_HALF_UP, ROUND_DOWN, ROUND_HALF_EVEN][int(c) - 1]
    # 根据指令符 c 的类型设置 decimal 舍入规则， UP、DOWN、EVEN 分别对应四舍五入、截断、四舍六入五成双
    tmp = Decimal(num).quantize(Decimal('0.' + '0' * d))  # 按格式要求计算出需要输出的数值
    print(f"{'-' if Decimal(tmp) < 0 else ''}{abs(tmp)}")  # 处理数值的符号，按格式要求打印输出结果
```

```python
def fun1(num,len_D):
    """09_16_homework.ipynb
    PAT-1123 舍入
    四舍五入
    """
    num_str = str(num)
    if '.' not in num_str:
        return num_str + '.' + '0'*len_D
    int_part, dec_part = num_str.split('.')
    if len(dec_part) <= len_D:
        return num_str + '0'*(len_D-len(dec_part))
    else:
        if int(dec_part[len_D]) >= 5:
            dec_part = list(dec_part[:len_D])
            for i in range(len_D-1, -1, -1):
                if dec_part[i] == '9':
                    dec_part[i] = '0'
                else:
                    dec_part[i] = str(int(dec_part[i]) + 1)
                    break
            else:
                int_part = str(int(int_part) + 1)
            dec_part = ''.join(dec_part)
        else:
            dec_part = dec_part[:len_D]
        return int_part + '.' + dec_part

def fun2(num, len_D):
    """09_16_homework.ipynb
    PAT-1123 截断
    直接截断，不进位
    """
    num_str = str(num)
    if '.' not in num_str:
        return num_str + '.' + '0'*len_D
    int_part, dec_part = num_str.split('.')
    if len(dec_part) <= len_D:
        return num_str + '0'*(len_D-len(dec_part))
    else:
        dec_part = dec_part[:len_D]
        return int_part + '.' + dec_part
    
def fun3(num, len_D):
    """09_16_homework.ipynb
    PAT-1123 四舍六入五成双
    遇到5时采用银行家舍入法
    """
    num_str = str(num)
    if '.' not in num_str:
        return num_str + '.' + '0'*len_D
    int_part, dec_part = num_str.split('.')
    if len(dec_part) <= len_D:
        return num_str + '0'*(len_D-len(dec_part))
    else:
        if int(dec_part[len_D]) > 5:
            dec_part = list(dec_part[:len_D])
            for i in range(len_D-1, -1, -1):
                if dec_part[i] == '9':
                    dec_part[i] = '0'
                else:
                    dec_part[i] = str(int(dec_part[i]) + 1)
                    break
            else:
                int_part = str(int(int_part) + 1)
            dec_part = ''.join(dec_part)
        elif int(dec_part[len_D]) < 5:
            dec_part = dec_part[:len_D]
        else: # == 5
            if len_D == 0: # 整数部分
                if int(int_part[-1]) % 2 == 0: # 偶数
                    dec_part = '0'*len_D
                else: # 奇数
                    int_part = str(int(int_part) + 1)
                    dec_part = '0'*len_D
            else: # 小数部分
                if len(dec_part) >= len_D + 1 and any([int(i) != 0 for i in dec_part[len_D+1:]]): # 5后面还有非0数字
                    # 直接进位
                    dec_part = list(dec_part[:len_D])
                    for i in range(len_D-1, -1, -1):
                        if dec_part[i] == '9':
                            dec_part[i] = '0'
                        else:
                            dec_part[i] = str(int(dec_part[i]) + 1)
                            break
                    else:
                        int_part = str(int(int_part) + 1)
                    dec_part = ''.join(dec_part)
                else: # 5后面没有非0数字
                    if int(dec_part[len_D-1]) % 2 == 0: # 偶数
                        dec_part = dec_part[:len_D]
                    else: # 奇数
                        dec_part = list(dec_part[:len_D])
                        for i in range(len_D-1, -1, -1):
                            if dec_part[i] == '9':
                                dec_part[i] = '0'
                            else:
                                dec_part[i] = str(int(dec_part[i]) + 1)
                                break
                        else:
                            int_part = str(int(int_part) + 1)
                        dec_part = ''.join(dec_part)
        return int_part + '.' + dec_part
    
if __name__ == "__main__":
    N, D = map(int, input().split())
    for _ in range(N):
        op, num = input().split()
        if op == '1':
            print(fun1(num, D))  # 直接传字符串，避免浮点数精度问题
        elif op == '2':
            print(fun2(num, D))
        elif op == '3':
            print(fun3(num, D))
```
