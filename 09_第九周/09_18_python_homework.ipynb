#### PAT-1123 舍入

```python
from decimal import *
 
 
# 获取输入数据
n, d = map(int, input().split())  # 获取首行输入
 
# 处理结果并输出
context = getcontext()  # 获取 decimal 模块的设置
context.prec = 300  # 设置 decimal 精度
for _ in range(n):  # 遍历行
    c, num = input().split()  # 获取该行的输入数据
    context.rounding = [ROUND_HALF_UP, ROUND_DOWN, ROUND_HALF_EVEN][int(c) - 1]
    # 根据指令符 c 的类型设置 decimal 舍入规则， UP、DOWN、EVEN 分别对应四舍五入、截断、四舍六入五成双
    tmp = Decimal(num).quantize(Decimal('0.' + '0' * d))  # 按格式要求计算出需要输出的数值
    print(f"{'-' if Decimal(tmp) < 0 else ''}{abs(tmp)}")  # 处理数值的符号，按格式要求打印输出结果
```
