````python
import re

# 读取26个字母的点阵图案，每个字母7行
words = []
for _ in range(26):
    letter_lines = []
    for _ in range(7):
        line = input().strip()
        letter_lines.append(line)
    words.append(letter_lines)

# 读取目标字符串并提取字母单词
target = input().strip()
word_list = re.findall(r'[A-Z]+', target)

# 输出点阵图案
for idx, word in enumerate(word_list):
    # 输出每个单词的7行点阵
    for i in range(7):
        # 收集当前行每个字母的点阵行
        lines_to_print = []
        for char in word:
            index = ord(char.upper()) - ord('A')
            lines_to_print.append(words[index][i])
        # 用空格连接每个字母的点阵行，避免末尾多余空格
        print(' '.join(lines_to_print))
    # 如果不是最后一个单词，输出空行分隔
    if idx < len(word_list) - 1:
        print()
````

#### PAT-1111 对称日期

````python
input = make_input(
    """
    5
    Feb 2, 2020
    Mar 7, 2020
    Oct 10, 101
    Nov 21, 1211
    Dec 29, 1229
    """
)

words_to_num = {'Jan':'01', 'Feb':'02', 'Mar':'03', 'Apr':'04', 'May':'05', 'Jun':'06',
                'Jul':'07', 'Aug':'08', 'Sep':'09', 'Oct':'10', 'Nov':'11', 'Dec':'12'}

N = int(input())
for _ in range(N):

    # 步骤1：解析输入日期
    month_day_part, year = input().split(',')

    # 步骤2：处理年份（去空格，补零）
    year = year.strip().zfill(4)

    # 步骤3：处理月日部分
    month_day_list = month_day_part.split()
    month_day_list[0] = words_to_num[month_day_list[0]]  # 月份转数字
    month_day_list[1] = month_day_list[1].zfill(2)      # 日期补零

    # 步骤4：拼接完整日期字符串
    month_day_str = ''.join(month_day_list)
    formatted_date = year + month_day_str

    # 步骤5：检查对称性
    if formatted_date == formatted_date[::-1]:
        print(f'Y {formatted_date}')
    else:
        print(f'N {formatted_date}')
````
