### 1. 描述什么是监督学习、非监督学习

*   **监督学习**：机器从**带有标签**的训练数据中学习，并尝试建立一个模型，以便对新的、未见过的数据做出预测或决策。
    *   **核心**：数据有“标准答案”（标签）。
    *   **例子**：给定一堆猫和狗的图片（特征），并且每张图片都标明了是“猫”还是“狗”（标签）。模型学习后，给定一张新的动物图片，它能预测出是猫还是狗。
    *   **主要任务**：分类（预测离散标签）和回归（预测连续值）。

*   **非监督学习**：机器从**没有标签**的数据中寻找内在结构或模式。
    *   **核心**：数据没有“标准答案”，让机器自己去发现数据中的故事。
    *   **例子**：给定一堆新闻文章，没有任何分类标签。模型可以自动将这些文章聚类成不同的主题（如体育、科技、财经）。
    *   **主要任务**：聚类、降维、关联规则学习等。

### 2. 监督学习算法一般有哪些步骤

1.  **数据收集与准备**：收集带有标签的数据集。
2.  **数据预处理与特征工程**：清洗数据（处理缺失值、异常值）、进行特征选择、特征缩放、特征编码等，以更好地表示数据。
3.  **数据集划分**：将数据集随机分为**训练集**（用于模型学习）、**验证集**（用于调参和模型选择）和**测试集**（用于最终评估模型性能）。
4.  **选择模型**：根据问题类型（分类/回归）选择合适的算法（如决策树、SVM、神经网络等）。
5.  **训练模型**：将训练集输入到算法中，算法通过优化代价函数来学习模型参数。
6.  **模型评估与调优**：使用验证集评估模型表现，并通过调整超参数（如学习率、树深度等）来优化模型，防止过拟合或欠拟合。
7.  **模型预测**：使用最终优化好的模型在**测试集**上进行预测，评估其泛化到新数据的能力。
8.  **模型部署**：将训练好的模型应用到实际生产环境中进行实时预测。

### 3. 分类网络和回归的主要区别

| 特性 | **分类** | **回归** |
| :--- | :--- | :--- |
| **预测目标** | **离散的**类别或标签（如：猫/狗/鱼；是/否） | **连续的**数值（如：房价、温度、销量） |
| **输出** | 类别、概率分布 | 具体的数值 |
| **算法目标** | 找到决策边界 | 找到最佳拟合线/曲线 |
| **评估指标** | 准确率、精准率、召回率、F1-score、AUC | 均方误差（MSE）、平均绝对误差（MAE）、R² |
| **例子** | 图像识别、垃圾邮件检测 | 房价预测、股票趋势分析 |

### 4. 列举常用分类算法，并解释说明

*   **逻辑回归**：虽然名字里有“回归”，但它是一种广泛使用的**分类**算法（通常是二分类）。它通过Sigmoid函数将线性回归的输出映射到(0,1)之间，解释为属于某个类的概率。
*   **决策树**：一种树形结构模型，通过一系列 if-else 问题对数据进行分割。它易于理解和解释。
*   **随机森林**：**集成学习**算法。通过构建多棵决策树并进行投票（“森林”），其最终预测结果由所有树的结果共同决定。它通过“集体智慧”降低了单棵决策树过拟合的风险，通常非常强大。
*   **支持向量机**：寻找一个**超平面**（在二维中就是一条线），以最大间隔将不同类别的数据点分开。对于非线性问题，可以使用核技巧映射到高维空间进行分割。
*   **K-近邻**：一种“懒惰学习”算法。对于一个新数据点，查看它在特征空间中最近的K个邻居，通过这K个邻居的多数投票来决定该点的类别。
*   **朴素贝叶斯**：基于贝叶斯定理，并假设特征之间相互独立（“朴素”）。虽然这个假设在现实中很难成立，但它在文本分类等领域非常有效。

### 5. 正确率、精准率、召回率的公式

首先定义混淆矩阵：
*   **真正例**：实际为正，预测为正。
*   **假正例**：实际为负，预测为正。
*   **真负例**：实际为负，预测为负。
*   **假负例**：实际为正，预测为负。

*   **准确率**：所有预测中，预测正确的比例。
    `Accuracy = (TP + TN) / (TP + TN + FP + FN)`

*   **精准率**：**预测为正例**的样本中，真正为正例的比例。关注**预测的准确性**。
    `Precision = TP / (TP + FP)`

*   **召回率**：**实际为正例**的样本中，被成功预测为正例的比例。关注**查全率**，即不要漏掉正例。
    `Recall = TP / (TP + FN)`

### 6. 列举常用的回归类算法并简单描述

*   **线性回归**：最基础的算法，试图找到一条直线（或超平面）来拟合数据，使得预测值和真实值之间的误差最小。
*   **多项式回归**：线性回归的扩展，通过添加特征的高次项来拟合非线性关系。
*   **决策树回归**：与分类树类似，但每个叶子节点存储的是一个连续值（通常是该叶子节点内样本目标值的平均值），而不是类别标签。
*   **随机森林回归**：集成多棵决策回归树，最终的预测结果是所有树预测值的**平均**。
*   **支持向量回归**：SVM的回归版本。其目标不是最大化间隔，而是定义一个“间隔带”，尽可能让所有数据点都落在这个间隔带内，同时最大化间隔带的宽度。

### 7. 回归模型算法的常用评估方法

*   **均方误差**：最常用的指标，衡量预测值与真实值之间差异的平方的平均值。对大的错误有更大的惩罚。
    `MSE = (1/n) * Σ(真实值 - 预测值)²`
*   **均方根误差**：MSE的平方根。它的单位与预测目标相同，更容易解释。
    `RMSE = √(MSE)`
*   **平均绝对误差**：衡量预测值与真实值之间绝对差异的平均值。对异常值不如MSE敏感。
    `MAE = (1/n) * Σ|真实值 - 预测值|`
*   **R² 决定系数**：表示模型能够解释的目标变量方差的比例。取值范围一般为[0,1]，越接近1说明模型拟合得越好。

### 8. 解释代价函数

**代价函数**（或称**损失函数**）是一个衡量**模型预测值**与**真实值**之间差异程度的函数。它是模型参数的函数。**机器学习模型训练的核心目标就是找到一组参数，使得代价函数的值最小化**。

*   **对于回归**：常用**均方误差**作为代价函数。
*   **对于分类**：常用**交叉熵损失**作为代价函数。

### 9. 解释一下熵的概念及应用场景

*   **概念**：在信息论中，熵是衡量**信息不确定性**或**混乱度**的指标。一个系统越不确定、越混乱，其熵就越大。
    *   **信息熵公式**：`H(X) = -Σ P(xᵢ) · log₂P(xᵢ)`，其中P(xᵢ)是事件xᵢ发生的概率。
    *   **直观理解**：一件事发生的概率越小，它发生时带来的信息量就越大。熵是所有可能事件的信息量的期望值。

*   **应用场景**：
    1.  **决策树算法**：用于计算**信息增益**。选择哪个特征进行分割的标准就是看用这个特征分割后，数据集的熵减少得最多（信息增益最大）。
    2.  **数据压缩**：熵定义了数据压缩的极限。熵越大，可压缩的空间越小。
    3.  **机器学习正则化**：在某些模型中，可以通过最大化熵来避免模型过于自信地做出预测，鼓励更均匀的概率分布，起到正则化作用。

### 10. 描述CART树

*   **CART**：**分类与回归树**的缩写。它是一种构建二叉决策树的算法。
*   **特点**：
    1.  **二叉树**：每个节点只分裂成两个子节点（即判断条件为“是/否”）。
    2.  **分裂标准**：
        *   **对于分类树**：使用**基尼不纯度**最小化作为分裂标准。基尼系数衡量一个随机选中的样本在子集中被错分类的概率。
        *   **对于回归树**：使用**均方误差**最小化作为分裂标准。选择那个能使分裂后子节点MSE总和最小的特征和阈值。
    3.  **输出**：
        *   分类树：叶子节点输出一个类别。
        *   回归树：叶子节点输出一个连续值（通常是该叶子节点内样本目标值的平均值）。
*   **过程**：递归地将特征空间划分为矩形区域，并在每个区域上做一个简单的预测。

### 11. 描述xgboost相关知识

*   **XGBoost**：**e**X**t**reme **G**radient **Boost**ing，是一种高效的、可扩展的梯度提升决策树算法。
*   **核心思想**：
    1.  **集成学习**：属于Boosting家族，通过**加法模型**（顺序地、串行地）训练多棵弱决策树（通常是CART树），每一棵树都在学习前一棵树的残差。
    2.  **梯度提升**：使用梯度下降算法来最小化自定义的损失函数，指导新树的生成。
*   **主要特点/优点**：
    1.  **正则化**：在目标函数中同时加入了**L1（Lasso）和 L2（Ridge）正则化项**，有效控制模型复杂度，防止过拟合。
    2.  **并行处理**：虽然树是串行生成的，但特征排序和分裂点查找的过程可以并行化，训练速度极快。
    3.  **处理缺失值**：算法内置了自动处理缺失值的机制。
    4.  **剪枝策略**：采用**深度优先**的“后剪枝”策略，更不容易过拟合。
    5.  **灵活性**：支持自定义损失函数（只要二阶可导）。
*   **应用**：因其出色的性能和速度，在各类数据科学竞赛和工业界中都非常流行。

### 12. 分类模型评估常用方法

1.  **混淆矩阵**：基础表格，直接给出TP, FP, TN, FN的值。
2.  **准确率**：最简单直观的指标，但在数据不平衡时可能失效。
3.  **精准率 & 召回率**：更细致的评估，尤其关注正例的预测情况。
4.  **F1-Score**：精准率和召回率的**调和平均数**，`F1 = 2 * (Precision * Recall) / (Precision + Recall)`，用于在两者间寻求平衡。
5.  **ROC曲线与AUC值**：
    *   **ROC曲线**：描绘了当分类器的阈值变化时，**真正例率** 和 **假正例率** 的变化关系。
    *   **AUC**：ROC曲线下的面积。AUC值越接近1，模型性能越好。它衡量的是模型的**排序能力**（将正样本排在负样本前面的能力），而不仅仅是预测的绝对准确性。
6.  **PR曲线**：当数据极度不平衡时，精准率-召回率曲线比ROC曲线更能反映模型性能。

### 13. 解释k折交叉验证？它的主要优点和缺点

*   **解释**：一种重采样技术，用于评估模型的泛化能力。
    1.  将原始数据集随机平均分成**K个互斥的子集**（称为“折”）。
    2.  每次轮流使用其中**1折**作为测试集，剩余的**K-1折**作为训练集，进行模型训练和评估。
    3.  重复这个过程K次，确保每折数据都有一次被用作测试集。
    4.  最终得到K个性能评估结果（如K个准确率），取其**平均值**作为模型最终的性能指标。

*   **优点**：
    1.  **有效减少过拟合**：充分利用了所有数据，既能用于训练也能用于测试，评估结果更可靠。
    2.  **更准确的评估**：比简单的留出法（一次划分训练/测试集）更能反映模型的真实性能。

*   **缺点**：
    1.  **计算成本高**：需要训练K个模型，当数据集很大或模型很复杂时，计算开销是原始方法的K倍。

### 14. 解释什么是过拟合和欠拟合问题

*   **欠拟合**：
    *   **表现**：模型在**训练集**和**测试集**上的表现都很差。
    *   **原因**：模型过于简单，无法捕捉数据中的潜在规律和特征。
    *   **比喻**：**“不好好学习”**。考试前没复习好，所以考试（训练和测试）都考砸了。

*   **过拟合**：
    *   **表现**：模型在**训练集**上表现非常好（甚至完美），但在**测试集**上表现很差。
    *   **原因**：模型过于复杂，把训练数据中的**噪声**和**特例**都当成了普遍规律来学习，导致泛化能力下降。
    *   **比喻**：**“死记硬背，不会举一反三”**。把习题集（训练集）的答案全背下来了，但遇到新题型（测试集）就不会做了。

### 15. 列举解决过拟合的方法

1.  **获取更多数据**：最有效的方法。
2.  **数据增强**：对现有数据进行变换，创造更多“新”数据（主要用于图像等领域）。
3.  **降低模型复杂度**：例如，减少决策树的深度、减少神经网络的层数或神经元数量。
4.  **正则化**：在损失函数中添加惩罚项（如L1/L2正则化），限制参数的大小，使模型变得更平滑。
5.  **集成方法**：使用Bagging（如随机森林）或Boosting（如XGBoost）来降低方差。
6.  **Dropout**：在神经网络训练中随机“丢弃”一部分神经元，防止神经元之间产生过强的依赖。
7.  **早停**：在训练迭代过程中，监控验证集性能，当性能不再提升时提前停止训练。

### 16. 描述对信息增益的理解

*   **理解**：信息增益是决策树（ID3算法）选择分裂特征时使用的标准。它衡量的是使用某个特征进行分割后，**数据集不确定性（熵）减少的程度**。信息增益越大，意味着使用该特征分割后，数据集的纯度提升得越多，分类效果越好。
*   **计算**：`信息增益 = 分裂前父节点的熵 - （所有子节点的熵的加权平均）`
*   **缺点**：对可取值数目较多的特征有偏好（例如“ID”特征，每个样本一个值，分裂后熵为0，增益最大，但这毫无意义）。为了克服这一点，后来有了**增益率**（C4.5算法）等改进标准。

### 17. 剪枝处理的作用及策略？

*   **作用**：剪枝是决策树中为了**防止过拟合**的核心手段。通过主动剪掉树的一些分支（子树或叶子节点），来降低模型的复杂度，提高其泛化能力。
*   **策略**：
    1.  **预剪枝**：在树生成的过程中，在每个节点分裂前进行估计。如果当前节点的分裂不能带来模型泛化性能的提升（如通过验证集判断），就停止分裂，将其标记为叶子节点。**优点**：降低过拟合风险，减少训练时间。**缺点**：可能带来欠拟合风险，因为分裂可能提前终止。
    2.  **后剪枝**：先完整地生成一棵决策树，然后自底向上地对非叶子节点进行考察。如果将该节点对应的子树替换为叶子节点能带来模型泛化性能的提升，则将该子树替换为叶子节点。**优点**：泛化性能通常优于预剪枝。**缺点**：训练时间开销大。

### 18. 解释AUC/ROC

*   **ROC曲线**：**接收者操作特征曲线**。它是一个二维曲线图，用于可视化分类模型在不同分类阈值下的性能。
    *   **横轴**：**假正例率**（False Positive Rate, FPR），即所有负样本中被错判为正的比例。`FPR = FP / (FP + TN)`
    *   **纵轴**：**真正例率**（True Positive Rate, TPR，即召回率），即所有正样本中被正确判为正的比例。`TPR = TP / (TP + FN)`
    *   曲线上的每一个点对应一个特定的分类阈值。

*   **AUC**：**ROC曲线下的面积**。
    *   **物理意义**：随机选取一个正样本和一个负样本，分类器将正样本排序高于负样本的**概率**。
    *   **取值范围**：[0, 1]。
    *   **解读**：
        *   AUC = 0.5：模型没有区分能力，相当于随机猜测。
        *   AUC > 0.5：优于随机猜测。越接近1，模型性能越好。
        *   AUC < 0.5：通常意味着模型有问题，把正负样本弄反了。
    *   **优点**：对类别不平衡不敏感，评价的是模型的排序能力。

### 19. 描述对支持向量机的了解

SVM是一种强大的**二分类**模型，它的核心思想是寻找一个**最优超平面**来对数据进行分割，并且这个超平面要距离两个类别的**最近数据点（支持向量）最远**，这个距离被称为**间隔**。最大化间隔可以使模型的泛化误差最小化。

*   **核技巧**：对于线性不可分的数据，SVM通过**核函数**将原始数据映射到高维特征空间，从而在高维空间中找到一个线性的最优超平面。常用的核函数有线性核、多项式核、高斯径向基核。
*   **软间隔**：为了处理噪声和 outliers，引入“软间隔”概念，允许一些样本点落在间隔内甚至被错分，通过一个惩罚参数C来平衡“间隔最大化”和“分类误差最小化”。
*   **支持向量**：最终模型只由那些位于间隔边界上的少数样本点（支持向量）所决定，这使得SVM模型非常高效和节省内存。

### 20. 描述均方误差、绝对值误差、f1_score公式

*   **均方误差**：预测值和真实值之**差的平方**的平均值。`MSE = (1/n) * Σ(y_i - ŷ_i)²`
*   **平均绝对误差**：预测值和真实值之**差的绝对值**的平均值。`MAE = (1/n) * Σ|y_i - ŷ_i|`
*   **F1-Score**：**精准率（Precision）** 和 **召回率（Recall）** 的**调和平均数**。
    `F1 = 2 * (Precision * Recall) / (Precision + Recall)`
    `= (2 * TP) / (2 * TP + FP + FN)`

2. 考虑到可以找出所有符合长度的素数，但貌似麻烦，可以找出所有符合长度的连续子串，判断是否是素数。
3. 直接使用**字符串的切片**，拿出符合长的的连续子串，再转数字进行判断

```python
def is_prime(n):
    """判断一个数是否为素数"""
    if n <= 1:
        return False
    if n <= 3:
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True


if __name__ == "__main__":
    L, K = map(int, input().split())
    N = input()
    begin = 0
    end = K
    while end <= L:
        if is_prime(int(N[begin:end])):
            print(N[begin:end])
            break
        begin += 1
        end += 1
    else:
        print("404")

```

#### PAT-1095
![](./images/1095.png)

1. 三个统计类型，可以按照不同的统计类型，创建以不同键来索引的字典。

```python
# 简化但高效的版本
def solve_pat_1095_fast():
    """
    简化但高效的版本，平衡可读性和性能
    """
    import sys
    input = sys.stdin.readline
    
    N, M = map(int, input().split())
    
    # 直接构建所需的数据结构
    level_data = {'A': [], 'B': [], 'T': []}
    site_data = {}
    date_data = {}
    
    # 一次性读取和分类
    for _ in range(N):
        line = input().split()
        exam_id, score = line[0], int(line[1])
        
        # 直接分类存储
        level = exam_id[0]
        if level in level_data:
            level_data[level].append((exam_id, score))
        
        site = exam_id[1:4]
        if site not in site_data:
            site_data[site] = []
        site_data[site].append(score)
        
        date = exam_id[4:10]
        if date not in date_data:
            date_data[date] = []
        date_data[date].append(exam_id[1:4])
    
    # 预排序级别数据
    for level in level_data:
        level_data[level].sort(key=lambda x: (-x[1], x[0]))
    
    # 预计算站点统计
    site_stats = {site: (len(scores), sum(scores)) 
                  for site, scores in site_data.items()}
    
    # 预计算日期统计
    from collections import Counter
    date_stats = {}
    for date, sites in date_data.items():
        counter = Counter(sites)
        sorted_sites = sorted(counter.items(), key=lambda x: (-x[1], x[0]))
        date_stats[date] = sorted_sites
    
    # 快速查询处理
    for case_num in range(1, M + 1):
        query = input().split()
        query_type, command = int(query[0]), query[1]
        
        print(f"Case {case_num}: {query_type} {command}")
        
        if query_type == 1:
            data = level_data.get(command, [])
            if data:
                for exam_id, score in data:
                    print(f"{exam_id} {score}")
            else:
                print("NA")
        
        elif query_type == 2:
            if command in site_stats:
                count, total = site_stats[command]
                print(f"{count} {total}")
            else:
                print("NA")
        
        elif query_type == 3:
            if command in date_stats:
                for site, count in date_stats[command]:
                    print(f"{site} {count}")
            else:
                print("NA")

# 使用最快的版本
if __name__ == "__main__":
    solve_pat_1095_fast()
```
