#### PAT-1096 大美数判断

**大美数定义：**
- 若正整数 N 可以**整除**它的 **4 个不同正因数之和**，则称这样的正整数为"大美数"。
- 换句话说：存在N的4个不同正因数 a, b, c, d，使得 N % (a + b + c + d) == 0。

**算法思路：**
1. **找因数**：找出N的所有正因数。
2. **排序正因数列表**：对因数列表进行排序，使得列表中数字的大小顺序是从小到大。
2. **组合选择**：使用四个有顺序的指针，从列表里取出从小到大四个正因数。
3. **整除判断**：检查N是否能被这4个因数的和整除。
4. **移动指针**：移动最后一个指针，并判断新的组合之和是否大于N，若大于，则将前一个指针右移，当前指针移到上一指针的右侧

**示例分析：**
- 数字 12 的因数：[1, 2, 3, 4, 6, 12]
- 选择4个因数：(1, 2, 3, 6)，和为 12
- 12 ÷ 12 = 1，能整除，所以 12 是大美数

**优化策略：**
- 提前终止：如果最小4个因数的和都大于N，则不可能是大美数
- 跳过过大组合：如果当前组合和超过N，跳过后续更大的组合

```python
def is_big_beautiful_number(N):
    """
    判断正整数N是否为"大美数"
    大美数定义：正整数N可以整除它的4个不同正因数之和
    
    Args:
        N (int): 待判断的正整数
    
    Returns:
        bool: 如果是大美数返回True，否则返回False
    """
    if N <= 0:
        return False
    
    # 第一步：找到N的所有正因数
    def get_divisors(num):
        """获取一个数的所有正因数"""
        divisors = []
        for i in range(1, int(num**0.5) + 1):
            if num % i == 0:
                divisors.append(i)
                if i != num // i:  # 避免重复添加平方根
                    divisors.append(num // i)
        return sorted(divisors)
    
    divisors = get_divisors(N)
    
    # 检查是否有至少4个不同的正因数
    if len(divisors) < 4:
        return False
    
    # 第二步：检查是否存在4个不同正因数的组合，其和能被N整除
    from itertools import combinations
    
    # 遍历所有4个因数的组合
    for four_divisors in combinations(divisors, 4):
        divisors_sum = sum(four_divisors)
        # 检查N是否能整除这4个因数的和
        if N % divisors_sum == 0:
            return True
    
    return False

# 优化版本：提前剪枝，提高效率
def is_big_beautiful_number_optimized(N):
    """
    优化版本的大美数判断函数
    使用提前剪枝来提高效率
    """
    if N <= 0:
        return False
    
    # 获取所有正因数
    divisors = []
    for i in range(1, int(N**0.5) + 1):
        if N % i == 0:
            divisors.append(i)
            if i != N // i:
                divisors.append(N // i)
    
    divisors.sort()
    
    if len(divisors) < 4:
        return False
    
    # 优化：提前剪枝
    # 如果最小的4个因数的和都大于N，则不可能是大美数
    min_sum = sum(divisors[:4])
    if min_sum > N:
        return False
    
    # 检查所有4个因数的组合
    from itertools import combinations
    
    for four_divisors in combinations(divisors, 4):
        divisors_sum = sum(four_divisors)
        # 如果和太大，后续组合的和会更大，可以跳过
        if divisors_sum > N:
            continue
        if N % divisors_sum == 0:
            return True
    
    return False

# 测试函数
def test_big_beautiful_numbers():
    """测试大美数函数"""
    print("测试大美数判断函数：")
    
    # 测试一些数字
    test_numbers = [12, 28, 30, 36, 60]
    
    for num in test_numbers:
        result = is_big_beautiful_number_optimized(num)
        divisors = []
        for i in range(1, int(num**0.5) + 1):
            if num % i == 0:
                divisors.append(i)
                if i != num // i:
                    divisors.append(num // i)
        divisors.sort()
        
        print(f"数字 {num}:")
        print(f"  所有因数: {divisors}")
        print(f"  是否为大美数: {result}")
        
        # 如果是大美数，找出一个满足条件的组合
        if result:
            from itertools import combinations
            for four_divisors in combinations(divisors, 4):
                divisors_sum = sum(four_divisors)
                if num % divisors_sum == 0:
                    print(f"  满足条件的4个因数: {four_divisors}, 和为: {divisors_sum}")
                    print(f"  {num} ÷ {divisors_sum} = {num // divisors_sum}")
                    break
        print()

# 运行测试
if __name__ == "__main__":
    test_big_beautiful_numbers()
```
