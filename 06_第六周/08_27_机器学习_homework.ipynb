#### PAT-1096 大美数判断

**大美数定义：**
- 若正整数 N 可以**整除**它的 **4 个不同正因数之和**，则称这样的正整数为"大美数"。
- 换句话说：存在N的4个不同正因数 a, b, c, d，使得 N % (a + b + c + d) == 0。

**算法思路：**
1. **找因数**：找出N的所有正因数。
2. **排序正因数列表**：对因数列表进行排序，使得列表中数字的大小顺序是从小到大。
2. **组合选择**：使用四个有顺序的指针，从列表里取出从小到大四个正因数。
3. **整除判断**：检查N是否能被这4个因数的和整除。
4. **移动指针**：移动最后一个指针，并判断新的组合之和是否大于N，若大于，则将前一个指针右移，当前指针移到上一指针的右侧

**示例分析：**
- 数字 12 的因数：[1, 2, 3, 4, 6, 12]
- 选择4个因数：(1, 2, 3, 6)，和为 12
- 12 ÷ 12 = 1，能整除，所以 12 是大美数

**优化策略：**
- 提前终止：如果最小4个因数的和都大于N，则不可能是大美数
- 跳过过大组合：如果当前组合和超过N，跳过后续更大的组合

```python
def is_big_beautiful_number(N):
    # 获取所有正因数
    divisors = []
    for i in range(1, int(N**0.5) + 1):
        if N % i == 0:
            divisors.append(i)
            if i != N // i:
                divisors.append(N // i)
    divisors = sorted(divisors)
    
    # 检查是否有至少4个不同的正因数
    if len(divisors) < 4:
        return False
    
    # 使用四重循环遍历所有可能的四因数组合
    n_divisors = len(divisors)
    for i in range(n_divisors):
        for j in range(i + 1, n_divisors):
            for k in range(j + 1, n_divisors):
                for l in range(k + 1, n_divisors):
                    total = divisors[i] + divisors[j] + divisors[k] + divisors[l]

                    if total % N == 0:
                        return True
                    
                    if divisors[i] + divisors[j] + divisors[k] + divisors[l+1] > N:
                        break

                if divisors[i] + divisors[j] + divisors[k+1] + divisors[k+2] > N:
                    break

            if divisors[i] + divisors[j+1] + divisors[k+2] + divisors[k+3] > N:
                break
    return False

if __name__ == "__main__":
    K = int(input())
    test_numbers = list(map(int, input().split()))
    for num in test_numbers:
        if is_big_beautiful_number(num):
            print('Yes')
        else:
            print('No')
```

#### PAT097
1. 给定一个 n×n 的整数矩阵。对任一给定的正整数 k < n，我们将矩阵的奇数行的元素整体向右依次平移 1、2、3、……、k个位置，平移空出的位置用整数 x 补。你需要计算出结果矩阵的每一列元素的和。
2. 用列表存储矩阵的每一行。
3. 生成循环列表，用于确定对每一行元素要移动多少。
4. 使用zip()函数与*解包操作，对二维列表（即矩阵）进行转置。

```python
n, k, fill = map(int, input().split())

# 读取矩阵
matrix = []
for i in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)

# 创建循环列表
def create_cycle_list(pattern, length):
    """创建任意模式的循环列表"""
    from itertools import cycle, islice
    return list(islice(cycle(pattern), length))
ccl = create_cycle_list([i for i in range(1,k+1)], n//2+1)

# 按要求变换矩阵
ccl_p = 0
for i in range(0,n,2):
    for j in range(ccl[ccl_p]):
        matrix[i].insert(0, fill)
        matrix[i].pop()

    ccl_p += 1

# 使用 zip(*matrix) 转置
transposed = list(zip(*matrix))

result = []
for i in transposed:
    sum_i = sum(i)
    result.append(str(sum_i))

print(' '.join(result))
```

```python
N = int(input())
up = map(int, input().split())
down = map(int, input().split())

high = min(up)
low = max(down)
if high > low:
    print("Yes", high - low)
else:
    print("No", low - high + 1)
```
