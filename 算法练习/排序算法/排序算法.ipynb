# 冒泡排序(Bubble Sort)
1. 冒泡排序的思路是，每次比较相邻的两个数，按照你想要的排序方式重新排序这两个数，并且将排好的数放在最后面

2. 第一层遍历的范围是0~n, 第二层遍历的范围是0~(n-i-1)

3. 从数组的第一个元素开始，比较**相邻的两个元素**

4. 如果顺序错误（前面的元素大于后面的元素），就交换它们

5. 每一轮遍历都会将当前未排序部分的最大元素"冒泡"到正确位置

6. 重复这个过程，直到整个数组有序

7. 可比喻为，捡了西瓜，丢了芝麻

```python
input_num = [3, 7, 4, 2, 8, 5, 1, 6]

def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(n-i-1):
            # 升序排列，若要降序，则将“>”改为“<”
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
        
        if not swapped:
            break
    return arr

print(bubble_sort(input_num))
```

# 选择排序(Selection Sort)
1. 将数组分为**已排序**和**未排序**两部分

2. 在**未排序部分**中找到最小元素

3. 将该最小元素与**未排序部分**的第一个元素交换

4. 这样已排序部分就增加了一个元素

5. 重复直到所有元素都排序

```python
input_num = [3, 7, 4, 2, 8, 5, 1, 6]

def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        # 记录当前位置，以当前位置为初始比较对象，遍历后面的数，与之比较，按排序要求，对其进行更新
        target_index = i
        for j in range(i, n):
            if arr[j] < arr[target_index]:
                target_index = j


        arr[i], arr[target_index] = arr[target_index], arr[i]

    return arr

print(selection_sort(input_num))
```

# 插入排序(Insertion Sort)
1. 将数组分为已排序和未排序两部分

2. 从未排序部分取出第一个元素

3. 在已排序部分从后向前扫描，找到合适的位置插入

4. 将大于该元素的所有元素向后移动一位

5. 插入元素到空出的位置

```python
# 我的版本
input_num = [3, 7, 4, 2, 8, 5, 1, 6]

def insertion_sort(arr):
    n = len(arr)
    for i in range(n-1):
        target_index = i+1

        for j in range(i,-1,-1):
            swapped = False
            if arr[target_index] < arr[j]:
                arr[target_index], arr[j] = arr[j], arr[target_index]
                target_index = j
                swapped = True
            if not swapped:
                break
    
    return arr

print(insertion_sort(input_num))
```

```python
# 经典版本
def insertion_sort_traditional(arr):
    n = len(arr)
    for i in range(1, n):
        # 取出未排序区域的第一个数
        key = arr[i]
        # 计算已排序区域的最大索引
        j = i - 1
        # 判断，按照已排序区域倒着遍历与key比较，不符合排序规则，则将arr[j]向后移一位
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        # 将key插入到空缺位置
        arr[j + 1] = key
    return arr
```

# 快速排序(Quick Sort)
1. 选择一个元素作为"基准"（pivot）

2. 重新排列数组，所有比基准小的放在左边，比基准大的放在右边

3. 递归地对左右两个子数组进行快速排序

4. 基准元素最终会位于其最终位置

```python
# 创建了新的数组空间
input_num = [3, 7, 4, 2, 8, 5, 1, 6]

def quick_sort(nums):
    n = len(nums)
    if n <= 1:
        return nums
    pivot = nums[0]
    left = [nums[i] for i in range(1, n) if nums[i] < pivot]
    right = [nums[i] for i in range(1, n) if nums[i] >= pivot]
    return quick_sort(left) + [pivot] + quick_sort(right)

print(quick_sort(input_num))
```
